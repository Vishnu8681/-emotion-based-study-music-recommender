## Interactive Plotting Example (Confusion Matrix)


```python
# Mock data mimicking test results for illustration
classes = ['anger', 'fear', 'joy', 'love', 'sadness', 'surprise']
y_test = np.random.choice(6, 500)
y_pred = np.random.choice(6, 500)

# Adjust mock to simulate high accuracy (diagonal dominance)
mask = np.random.rand(500) < 0.85
y_pred[mask] = y_test[mask]

cm = confusion_matrix(y_test, y_pred)
plt.figure(figsize=(7, 5))
sns.heatmap(cm, annot=True, fmt='d', cmap='Blues', xticklabels=classes, yticklabels=classes)
plt.title('Interactive Confusion Matrix')
plt.ylabel('True')
plt.xlabel('Predicted')
plt.show()
```
