# 📊 Exploratory Data Analysis (EDA) - HarmonyMind
This notebook explores the Kaggle emotions dataset used for training our text emotion classifier. We perform shape analysis, class distribution checks, text length distribution, and word cloud visualizations.


```python
import os
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from collections import Counter
import re
import nltk
from nltk.corpus import stopwords
nltk.download('stopwords')
```

## Load Dataset


```python
# Dynamic path resolution
project_root = os.path.dirname(os.getcwd())
data_path = os.path.join(project_root, 'datasets', 'train.txt')

print('Loading dataset from:', data_path)
df = pd.read_csv(data_path, sep=';', header=None, names=['text', 'emotion'])
print('Shape:', df.shape)
print('Columns:', df.columns)
df.head()
```

## Class Distribution Analysis


```python
emotion_counts = df['emotion'].value_counts()
print(emotion_counts)

plt.figure(figsize=(8, 5))
sns.barplot(x=emotion_counts.index, y=emotion_counts.values, palette='Purples_r')
plt.title('Emotion Label Frequencies')
plt.xlabel('Emotion')
plt.ylabel('Count')
plt.tight_layout()
plt.show()
```

## Sequence Length Analysis


```python
df['word_count'] = df['text'].apply(lambda x: len(str(x).split()))
print(df['word_count'].describe())

plt.figure(figsize=(8, 4))
sns.histplot(df['word_count'], bins=30, kde=True, color='indigo')
plt.title('Word Count Distribution in Text Inputs')
plt.xlabel('Word Count')
plt.ylabel('Frequency')
plt.show()
```

## Vocabulary Word Frequency


```python
stop_words = set(stopwords.words('english'))
all_words = []

for text in df['text']:
    words = re.sub(r'[^a-zA-Z\s]', '', str(text).lower()).split()
    all_words.extend([w for w in words if w not in stop_words])

word_freq = Counter(all_words).most_common(20)
print('Top 20 words (excluding stopwords):', word_freq)

# Plot top words
plt.figure(figsize=(10, 5))
words_df = pd.DataFrame(word_freq, columns=['word', 'count'])
sns.barplot(x='count', y='word', data=words_df, palette='viridis')
plt.title('Most Common Vocabulary Tokens')
plt.show()
```
